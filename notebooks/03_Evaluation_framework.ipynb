```python
import sys
from pathlib import Path
import pandas as pd

sys.path.append("../src")

from metrics import evaluate_pair, evaluate_predictions
from pdb_validation import validate_pdb
from predictor import dummy_model, predict_directory
from structure_io import prepare_split_subset, read_structure_as_pdb
```

## 2. Metric sanity check (ground truth vs itself)

```python
cif = sorted(Path("../data/structures/eval").rglob("*.cif"))[0]
gt = read_structure_as_pdb(cif)
validate_pdb(gt)

assert evaluate_pair(gt, gt) == {"RMSD": 0.0, "TM_score": 1.0, "GDT_TS": 1.0, "lDDT": 1.0}
```

## 3. Prepare references & generate predictions

```python
DATA_DIR = Path("../data")
SPLIT = "eval"
LIMIT = None

prepare_split_subset(
    cif_dir=DATA_DIR / "structures" / SPLIT,
    pdb_dir=DATA_DIR / "pdb" / SPLIT,
    fasta_dir=DATA_DIR / "fasta" / SPLIT,
    limit=LIMIT,
)
```

```python
predict_directory(
    fasta_dir=DATA_DIR / "fasta" / SPLIT,
    out_dir=DATA_DIR / "predictions" / SPLIT,
    model_fn=dummy_model,
    limit=LIMIT,
)
```

## 4. Evaluate predictions and export CSV

```python
results = evaluate_predictions(
    pred_dir=DATA_DIR / "predictions" / SPLIT,
    ref_dir=DATA_DIR / "pdb" / SPLIT,
    output_csv=DATA_DIR / "results" / SPLIT / "evaluation_metrics.csv",
)

pd.DataFrame(results).head(10)
```
