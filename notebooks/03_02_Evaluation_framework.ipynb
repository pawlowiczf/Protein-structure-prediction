## Metric tests

Both metrics take CA-only PDB text. A structure compared with itself must give a perfect score.

#### TM-score

```python
cif = sorted(Path("../data/structures/eval/af").glob("*.cif"))[0]
gt = read_structure_as_pdb(cif)  # .cif -> CA-only PDB text
n_residues = len(validate_pdb(gt))

result = tm_score(gt, gt)

# Identical structures - TM-score should be 1.0, RMSD 0.0.
print(result)

assert result["rmsd"] == 0.0
assert result["tm_score"] == 1.0
assert result["gdt_ts"] == 1.0
assert result["n_common"] == n_residues
```

### lDDT score
